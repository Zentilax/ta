```python
def split_by_bab(df):
    df_known_bab = df[df['BAB'] != 'UNKNOWN'].reset_index(drop=True)
    df_unknown_bab = df[df['BAB'] == 'UNKNOWN'].reset_index(drop=True)
    return df_known_bab, df_unknown_bab
```

```python
df_known, df_unknown = split_by_bab(df)
```

```python
df_unknown = df_unknown.drop(columns=['BAB'])
```
