```python
import os

# Define source and destination directories
src_dir = r"C:\Users\Aqil Anhein\Desktop\TA\kg"
dst_dir = r"C:\Users\Aqil Anhein\Desktop\TA\kg\csv"

# Make sure the destination folder exists
os.makedirs(dst_dir, exist_ok=True)

# Dictionary of DataFrames and their filenames
dfs = {
    "bab": bab,
    "bab_pasal": bab_pasal,
    "doc": doc,
    "doc_bab": doc_bab,
    "doc_pasal": doc_pasal,
    "pasal": pasal,
    "prev_fixed": prev_fixed
}

# Save each DataFrame as CSV
for name, df in dfs.items():
    csv_path = os.path.join(dst_dir, f"{name}.csv")
    df.to_csv(csv_path, index=False, encoding="utf-8-sig")

print("✅ All files saved successfully to:", dst_dir)
```
