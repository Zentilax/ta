```python
prev = pd.read_excel(r"C:\Users\Aqil Anhein\Desktop\TA\md\rels\prev.xlsx")
prev
```

```python
# 1. Remove rows where Start_id % 10 == 1
prev = prev[prev['Start_id'] % 1000 != 1].copy()

# 2. Add column End_id = Start_id - 1
prev['End_id'] = prev['Start_id'] - 1

# 3. Add column that checks if Start_id - End_id == 1
prev['Relation'] = prev.apply(lambda x: 'PREV' if x['Start_id'] - x['End_id'] == 1 else None, axis=1)

# 4. Keep only rows where Relation == 'PREV'
prev = prev[prev['Relation'] == 'PREV'].reset_index(drop=True)
prev
```
